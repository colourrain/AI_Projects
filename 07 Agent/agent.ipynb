```python
# 导入模型 豆包
import os
from langchain_openai import ChatOpenAI
#获取api_key
api_key = os.getenv('ARK_API_KEY')
model = ChatOpenAI(
    openai_api_base="https://ark.cn-beijing.volces.com/api/v3",
    openai_api_key=api_key,	# app_key
    model_name="doubao-seed-1-6-flash-250828",	# 您想使用的特定模型的名称或标识符。
    max_tokens=1000, #限制响应中的令牌总数，有效控制输出长度。
    temperature= 0.7,  #控制模型输出的随机性。值越高，响应越具创造性；值越低，响应越确定性。
    timeout=30, #模型的响应时间s
)
response = model.invoke("夏天适合吃什么水果")
```

```python
print(response.content)
```

```python
# 导入模型 ollma
from langchain_ollama import ChatOllama
ollama_model = ChatOllama(model="qwen2.5:0.5b")
response = ollama_model.invoke("介绍一下你自己")
print(response)
```

```python
# 创建模板 PromptTemplate
from langchain_core.prompts import PromptTemplate
template = """你是一个{role}，请用{style}风格回答问题:
{question}
"""
prompt = PromptTemplate.from_template(template)
# 填充模板
filled_prompt = prompt.format(role="学生", style="幽默", question="你叫什么名字？")
filled_prompt1 = prompt.invoke({"role": "数学老师", "style": "通俗易懂", "question": "勾股定理是什么？"})
ollama_model.invoke(filled_prompt1)
```

```python
# 创建模板 chatprompt
from langchain_core.prompts import ChatPromptTemplate  
system_template = "你是一个{role}，请用{style}风格回答问题"
user_template = "{question}"

chat_prompt = ChatPromptTemplate(
    messages=[
        {"role": "system", "content": system_template},
        {"role": "user", "content": user_template}
    ]
)   
filled_chat_prompt = chat_prompt.format(role="学生", style="幽默", question="你叫什么名字？")
msg = ollama_model.invoke(filled_chat_prompt)
print(msg)
```

```python
# 输出格式化 - 使用 PydanticOutputParser
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel, Field
from typing import List

# 定义输出的数据结构
class OutputFormat(BaseModel):
    """输出格式"""
    answer: str = Field(description="问题的答案")
    confidence: float = Field(description="答案的可信度，0-1之间")

# 创建parser
parser = PydanticOutputParser(pydantic_object=OutputFormat)
```
