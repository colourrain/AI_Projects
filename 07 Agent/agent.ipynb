```python
# 导入本地模型 ollma
from langchain_ollama import ChatOllama
ollama_model = ChatOllama(model="qwen2.5:3b")
response = ollama_model.invoke("介绍一下你自己")
print(response)
```

```python
# 创建模板 PromptTemplate
from langchain_core.prompts import PromptTemplate
template = """你是一个{role}，请用{style}风格回答问题:
{question}
"""
prompt = PromptTemplate.from_template(template)
# 填充模板
filled_prompt = prompt.format(role="学生", style="幽默", question="你叫什么名字？")
filled_prompt1 = prompt.invoke({"role": "数学老师", "style": "通俗易懂", "question": "勾股定理是什么？"})
ollama_model.invoke(filled_prompt1)
```

```python
# 创建模板 chatprompt
from langchain_core.prompts import ChatPromptTemplate  
system_template = "你是一个{role}，请用{style}风格回答问题"
user_template = "{question}"

chat_prompt = ChatPromptTemplate(
    messages=[
        {"role": "system", "content": system_template},
        {"role": "user", "content": user_template}
    ]
)   
filled_chat_prompt = chat_prompt.format(role="学生", style="幽默", question="你叫什么名字？")
msg = ollama_model.invoke(filled_chat_prompt)
print(msg)
```

```python
# 输出格式化 - 使用 PydanticOutputParser
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel, Field
from typing import List
from langchain_core.prompts import PromptTemplate

# 定义输出的数据结构
class OutputFormat(BaseModel):
    """输出格式"""
    answer: str = Field(description="问题的答案")
    confidence: float = Field(description="答案的可信度，0-1之间")

# 创建parser
parser = PydanticOutputParser(pydantic_object=OutputFormat)
# 定义模板，要求模型输出符合指定格式
template = """请回答以下问题，并以JSON格式输出，包含answer和confidence字段:
{question}
"""
prompt = PromptTemplate.from_template(template)
# 填充模板
filled_prompt = prompt.format(question="地球离太阳有多远？")
# 调用模型获取响应
response = ollama_model.invoke(filled_prompt)
# 解析模型输出
parsed_output = parser.parse(response.content)
print(parsed_output.answer)
print(parsed_output.confidence)
```

```python
# LLM调用链 - 使用管道操作符
# 创建链：prompt | model | parser
chain = prompt | ollama_model | parser
# 运行链
result = chain.invoke({"question": "地球离太阳有多远？"})
print(result)
```

```python
# 创建流式链 - 使用流式管道操作符
stream_chain = prompt | ollama_model
# 运行流式链    
for result in stream_chain.stream({"question": "地球离太阳有多远？"}):
    print(result.content, end='')  
```

```python
# trim the message
from langchain.messages import RemoveMessage
from langgraph.graph.message import REMOVE_ALL_MESSAGES
from langgraph.checkpoint.memory import InMemorySaver
from langchain.agents import create_agent, AgentState
from langchain.agents.middleware import before_model
from langgraph.runtime import Runtime
from langchain_core.runnables import RunnableConfig
from typing import Any


@before_model
def trim_messages(state: AgentState, runtime: Runtime) -> dict[str, Any] | None:
    """Keep only the last few messages to fit context window."""
    messages = state["messages"]

    if len(messages) <= 3:
        return None  # No changes needed

    first_msg = messages[0]
    recent_messages = messages[-3:] if len(messages) % 2 == 0 else messages[-4:]
    new_messages = [first_msg] + recent_messages

    return {
        "messages": [
            RemoveMessage(id=REMOVE_ALL_MESSAGES),
            *new_messages
        ]
    }

agent = create_agent(
    model=ollama_model,
    tools=[],
    middleware=[trim_messages],
    checkpointer=InMemorySaver(),
)

config: RunnableConfig = {"configurable": {"thread_id": "1"}}
agent.invoke({"messages": "hi, how are you!"}, config)
agent.invoke({"messages": "hi, my name is bob"}, config)
agent.invoke({"messages": "write a short poem about cats"}, config)
agent.invoke({"messages": "now do the same but for dogs"}, config)
final_response = agent.invoke({"messages": "what's my name?"}, config)

final_response["messages"][-1].pretty_print()
```

### 嵌入向量

```python
from langchain_ollama.embeddings import OllamaEmbeddings

embeddings = OllamaEmbeddings(model="qwen2.5:3b")
vector = embeddings.embed_documents(["很好","很差"])
print(f"向量的维度是：{len(vector[0])}")
print(f"向量：{vector[0]}") 
```

```python
from sklearn.linear_model import LinearRegression
test_vector = embeddings.embed_documents("非常好")
labels = [1,0] # 1 for positive, 0 for negative
model = LinearRegression()
clf = model.fit(vector, labels)
prediction = clf.predict(test_vector)
print("positive" if prediction[0] > 0.7 else "negative")
```

### 本地持久化

```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
# load the text
loader = TextLoader("./政策文件.txt", encoding="utf-8")

# read the text
documents = loader.load()
#中文分块
splitter = RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=0,separators=["\n\n", "\n", "。", "！", "？", ",", "，", " ", ""])
docs = splitter.split_documents(documents)
print(len(docs))
i=1
for doc in docs:
    print(f"page{i}:{doc.page_content}")
    i += 1
```

```python
from langchain_community.vectorstores import Chroma

db = Chroma.from_documents(docs, embeddings, collection_name="policy-documents", persist_directory="./chroma_db")

# 语义搜索
docs = db.similarity_search("政策文件中关于上班时间的内容有哪些？", k=3)
for doc in docs:
    print(doc.page_content) 
```
